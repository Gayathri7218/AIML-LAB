```python
#importing required libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.graph_objs as go
import plotly.express as px

#reading data from the dataset
df=pd.read_csv("/content/CAR DETAILS FROM CAR DEKHO.csv")

#Analyzing the dataset
df.info()
df.describe()
df.columns

#applying lambda function
df["name_2"] = df.name.apply(lambda x : ' '.join(x.split(' ')[:1]))

#without lambda function
#df["name_2"]=df["name"]

#visualizing the data
df.name_2.value_counts()
sns.countplot(data=df, x="name_2")
plt.xticks(rotation=90)
plt.xlabel("Name",fontsize=10,color="black")
plt.ylabel("Name",fontsize=10,color="black")
plt.title("NAME COUNT",color="black")
plt.show()
```

```python
#importing libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#load the dataset
df=pd.read_csv("/content/CAR DETAILS FROM CAR DEKHO.csv")

#describing data
df.info()

#identifying missing values
print(df.isnull().sum())

#describes all descriptive statistics
df.describe()

#displaying missing values in percentage
print(df.isnull().sum()/len(df)*100)

#visualyzing data
sns.histplot(data=df,x="selling_price",kde=True)#kernel distribution estimator
plt.xlabel("Selling Price")
plt.ylabel("Frequency")
plt.title("Distribution of selling price")
plt.show()

sns.histplot(data=df,x="year",kde=True)#kernel distribution estimator
plt.xlabel("year")
plt.ylabel("Frequency")
plt.title("Distribution of year")
plt.show()


```
